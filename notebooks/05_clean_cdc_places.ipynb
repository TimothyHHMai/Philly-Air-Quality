# 05 · Clean CDC PLACES

### Settings

```python
pd.set_option("display.max_columns", None)

RAW = Path("/Volumes/01-2TB/Philly-Air-Quality/raw")

# Philadelphia region: Pennsylvania, New Jersey, Delaware
STATE_CODES = ["42", "34", "10"]
STATE_ABBRS = ["PA", "NJ", "DE"]

PLACES = RAW / "cdc_places" / "PLACES_census_tract.csv"

CHUNK_SIZE = 500_000
```

### Functions

```python
def read_places():
    """Read the national PLACES file in chunks, keeping rows in STATE_ABBRS."""
    frames = []
    for chunk in pd.read_csv(PLACES, dtype=str, chunksize=CHUNK_SIZE):
        frames.append(chunk[chunk["StateAbbr"].isin(STATE_ABBRS)])
    return pd.concat(frames, ignore_index=True)
```

## Read data

```python
places = read_places()
```
