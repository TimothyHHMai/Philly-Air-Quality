# 04 · Clean NOAA HMS Smoke

### Settings

```python
pd.set_option("display.max_columns", None)

RAW = Path("/Volumes/01-2TB/Philly-Air-Quality/raw")

# Philadelphia region: Pennsylvania, New Jersey, Delaware
STATE_CODES = ["42", "34", "10"]
STATE_ABBRS = ["PA", "NJ", "DE"]

HMS = RAW / "hms_smoke"
```

### Functions

```python
def read_hms():
    """Read every daily smoke shapefile into one GeoDataFrame.

    file_date comes from the file name, so each polygon keeps the day it was mapped on.
    """
    frames = []
    for path in sorted(HMS.glob("[0-9]*/hms_smoke*.zip")):
        gdf = gpd.read_file(f"zip://{path}")
        gdf["file_date"] = path.stem.removeprefix("hms_smoke")
        frames.append(gdf)
    return pd.concat(frames, ignore_index=True)
```
