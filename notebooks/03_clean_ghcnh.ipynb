### Settings

```python
pd.set_option("display.max_columns", None)

RAW = Path("/Volumes/01-2TB/Philly-Air-Quality/raw")

# Philadelphia region: Pennsylvania, New Jersey, Delaware
STATE_CODES = ["42", "34", "10"]
STATE_ABBRS = ["PA", "NJ", "DE"]

GHCNH = RAW / "ghcnh"
```

### Functions

```python
def read_ghcnh():
    """Read every station-year .psv file into one DataFrame."""
    frames = [
        pd.read_csv(path, sep="|", dtype=str)
        for path in sorted(GHCNH.glob("[0-9]*/GHCNh_*.psv"))
    ]
    return pd.concat(frames, ignore_index=True)
```
