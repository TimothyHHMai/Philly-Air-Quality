# 01 · Clean EPA AQS AirData

Verified daily and hourly PM2.5 (88101, 88502) and ozone (44201), daily AQI by county, and the site and monitor lists, 2020–2026.

Source: [EPA AirData download files](https://aqs.epa.gov/aqsweb/airdata/download_files.html)

### Settings

```python
pd.set_option("display.max_columns", None)

RAW = Path("/Volumes/01-2TB/Philly-Air-Quality/raw")

# Philadelphia region: Pennsylvania, New Jersey, Delaware
STATE_CODES = ["42", "34", "10"]
STATE_ABBRS = ["PA", "NJ", "DE"]

AQS = RAW / "aqs_airdata"

CHUNK_SIZE = 500_000
```

### Functions

```python
def read_aqs(pattern):
    """Read every AQS zip matching pattern, keeping rows in STATE_CODES.

    The files are national, so they are read in chunks and filtered as they load.
    """
    frames = []
    for path in sorted(AQS.glob(pattern)):
        reader = pd.read_csv(path, dtype=str, chunksize=CHUNK_SIZE)
        for chunk in reader:
            frames.append(chunk[chunk["State Code"].isin(STATE_CODES)])
    return pd.concat(frames, ignore_index=True)
```
